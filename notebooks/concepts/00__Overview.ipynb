### **Overview**

Today’s project focuses on analysing a dataset containing **historical Nobel Prize winners**. The aim is to uncover **patterns, trends, and insights** about Nobel laureates over time and to better understand how the Nobel Prize reflects broader scientific, cultural, and geopolitical developments.

We will work with **real-world historical data**, applying a wide range of **data cleaning, exploration, and visualisation techniques** learned throughout previous days.

---

### **Historical Context**

* On **November 27, 1895**, **Alfred Nobel** signed his last will in Paris.
* After his death, the will caused controversy, as Nobel left much of his fortune to establish a prize.
* Nobel dictated that prizes should be awarded to those who, during the preceding year, **“have conferred the greatest benefit to humankind.”**
* Nobel Prizes are awarded annually in the following categories:

  * Chemistry
  * Physics
  * Physiology or Medicine
  * Literature
  * Economics
  * Peace
