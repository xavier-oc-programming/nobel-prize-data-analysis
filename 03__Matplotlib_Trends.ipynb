### **Challenge 2: Prize Share of Laureates over Time**

Key question:

* Are prizes increasingly being **shared among more laureates**?

#### **Approach**

* Compute the **yearly average prize share**
* Calculate a **5-year rolling average**
* Add a **secondary y-axis** to the existing chart
* Plot prize share alongside prize count

```python
yearly_avg_share = df_data.groupby('year').agg({'share_pct': 'mean'})
share_moving_average = yearly_avg_share.rolling(window=5).mean()
```
