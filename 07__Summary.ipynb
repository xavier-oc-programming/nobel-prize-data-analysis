## **5)— Learning Points & Summary**

### **Overview**

This project brought together a wide range of data-cleaning, aggregation, and visualisation techniques to analyse Nobel Prize data from multiple perspectives. The focus was not only on producing charts, but on understanding how different analytical choices shape the story the data tells.

---

### **Core Skills Reinforced**

* Identifying and investigating **NaN values** in real-world datasets.
* Converting **object and string data types** into numeric formats suitable for analysis.
* Using Plotly to create **bar charts** and **donut charts**.
* Applying **rolling averages** to smooth time-series data and reveal trends.
* Confident use of:

  * `.value_counts()`
  * `.groupby()`
  * `.merge()`
  * `.sort_values()`
  * `.agg()`

---

### **New Concepts Learned**

* Creating **choropleth maps** to visualise country-level data geographically.
* Building **segmented bar charts** to compare categories within totals.
* Using **sunburst charts** to represent hierarchical geographic structures.
* Analysing trends with Seaborn’s `.lmplot()` using:

  * `row` for faceting by category
  * `hue` for multi-category comparison
  * `lowess` to show smoothed best-fit trends
* Interpreting how **different visualisations of the same data** (e.g. box plots vs. time-series trends) can lead to different but complementary insights.
* Visualising distributions and descriptive statistics using **histograms** in Seaborn.

---

### **Key Insight**

The same dataset can tell very different stories depending on:

* Which columns are chosen
* How the data is grouped or aggregated
* Which visualisation technique is used

Understanding these choices is just as important as writing correct code.

---

### **Conclusion**

This project demonstrates how exploratory data analysis combines technical skill with analytical judgment. By approaching the data from multiple angles—geographic, temporal, categorical, and demographic—you gain a much deeper and more nuanced understanding than any single chart could provide.
