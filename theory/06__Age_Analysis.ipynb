## **6)— Unearthing Patterns in the Laureate Age at the Time of the Award**

### **Overview**

This section analyses how old Nobel laureates are when they receive their prize. The focus is on overall age distribution, identifying the youngest and oldest winners, examining trends over time, and comparing age patterns across prize categories.

---

### **Challenge 1: Calculating Age at the Time of the Award**

**Goal:** Add a new column representing each laureate’s age when they won the Nobel Prize.

**Approach:**

* Extract the birth year from the `birth_date` column.
* Subtract the birth year from the award year.
* Store the result in a new column called `winning_age`.

```python
birth_years = df_data.birth_date.dt.year
df_data['winning_age'] = df_data.year - birth_years
```

This column is used throughout the rest of the analysis.

---

### **Challenge 2: Oldest and Youngest Winners**

**Goal:** Identify the extremes in laureate age.

**Steps:**

* Use `nlargest()` to find the oldest winner.
* Use `nsmallest()` to find the youngest winner.

```python
df_data.nlargest(1, columns='winning_age')
df_data.nsmallest(1, columns='winning_age')
```

**Key findings:**

* The **oldest** Nobel laureate was **97 years old** at the time of the award.
* The **youngest** laureate was **17 years old**.
* These extremes highlight the wide range of ages at which major contributions are recognised.

---

### **Challenge 3: Descriptive Statistics and Age Distribution**

**Goal:** Understand the overall distribution of winning ages.

**Descriptive statistics:**

```python
df_data.winning_age.describe()
```

Key values:

* **Mean age:** ~60 years
* **Median age:** 60 years
* **75% of laureates** were younger than **69 years** at the time of the award

**Visualisation:**

* Use Seaborn’s `histplot()` to visualise the distribution.
* Experiment with different bin sizes (e.g. 10, 20, 30, 50) to see how the shape changes.

```python
plt.figure(figsize=(8, 4), dpi=200)
sns.histplot(
    data=df_data,
    x='winning_age',
    bins=30
)
plt.xlabel('Age')
plt.title('Distribution of Age on Receipt of Prize')
plt.show()
```

**Observation:**

* The distribution is roughly bell-shaped.
* Most prizes are awarded between ages **50 and 70**.

---

### **Challenge 4: Winning Age Over Time**

**Goal:** Determine whether Nobel laureates are receiving their awards later in life than before.

**Approach:**

* Plot award year against winning age.
* Use a regression plot with `lowess=True` to show a smoothed trend over time.

```python
plt.figure(figsize=(8, 4), dpi=200)
with sns.axes_style("whitegrid"):
    sns.regplot(
        data=df_data,
        x='year',
        y='winning_age',
        lowess=True,
        scatter_kws={'alpha': 0.4},
        line_kws={'color': 'black'}
    )
plt.show()
```

**Key insights:**

* From **1900–1940**, laureates were typically around **55 years old**.
* In recent decades, the average age has increased to **around 70 years**.
* The spread of ages has widened, with more very young and very old winners appearing in recent years.

---

### **Challenge 5: Age Differences Between Prize Categories**

**Goal:** Compare age distributions across Nobel Prize categories.

**Approach:**

* Use a box plot to visualise differences in median, spread, and outliers.

```python
plt.figure(figsize=(8, 4), dpi=200)
with sns.axes_style("whitegrid"):
    sns.boxplot(
        data=df_data,
        x='category',
        y='winning_age'
    )
plt.show()
```

**Insights:**

* **Physics** and **Chemistry** tend to have older winners.
* **Peace** shows a wider spread, including younger laureates.
* **Economics** is relatively stable compared to other categories.

---

### **Challenge 6: Age Trends Over Time by Category**

**Goal:** Examine how age trends differ across categories.

**Separate trends per category:**

```python
with sns.axes_style("whitegrid"):
    sns.lmplot(
        data=df_data,
        x='year',
        y='winning_age',
        row='category',
        lowess=True,
        aspect=2,
        scatter_kws={'alpha': 0.6},
        line_kws={'color': 'black'}
    )
plt.show()
```

**Combined trends using hue:**

```python
with sns.axes_style("whitegrid"):
    sns.lmplot(
        data=df_data,
        x='year',
        y='winning_age',
        hue='category',
        lowess=True,
        aspect=2,
        scatter_kws={'alpha': 0.5},
        line_kws={'linewidth': 5}
    )
plt.show()
```

**Key conclusions:**

* **Physics, Chemistry, and Medicine** show a strong upward age trend.
* **Physics** exhibits the steepest increase over time.
* **Peace** prizes trend younger over time.
* Different visualisations (box plots vs. trend lines) reveal different aspects of the data and should be used together.

---

### **Key Takeaways**

* Nobel Prizes are typically awarded around age 60.
* Laureates are receiving prizes later in life than in the early 20th century.
* Age trends vary significantly by category.
* Long-term trend analysis and distribution analysis provide complementary insights.
