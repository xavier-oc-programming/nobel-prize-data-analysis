## **2)—Plotly Bar & Donut Charts: Analyse Prize Categories & Women Winning Prizes**

### **Challenge 1: Asking the Right Questions**

A key part of data science is learning how to **ask meaningful questions** of the data. This process is open-ended and exploratory, and not every question will have a clear or immediate answer.

Examples of guiding questions:

* What percentage of Nobel laureates were women?
* How are prizes distributed across categories?
* Are some categories more gender-balanced than others?

These questions drive the analysis that follows.

---

### **Challenge 2: Gender Distribution of Nobel Prize Winners**

Create a **donut chart using Plotly** to compare how many prizes went to men versus women.

#### **Approach**

* Use `.value_counts()` on the `sex` column
* Use `px.pie()` with a hole parameter to create a donut chart
* Display percentages inside the chart

#### **Key Result**

* Since 1901, **only about 6.2% of Nobel Prize winners were women**
* Approximately **93.8%** of prizes went to men

---

### **Challenge 3: The First Women to Win a Nobel Prize**

Questions explored:

* Who were the first three female Nobel laureates?
* What categories did they win in?
* Where were they born?
* Were they affiliated with organisations?

#### **Findings**

The first three female Nobel Prize winners were:

* **Marie Curie** (Physics, 1903)
* **Bertha von Suttner** (Peace, 1905)
* **Selma Lagerlöf** (Literature, 1909)

All three were **individual laureates**, not organisations, and came from **European countries**.

---

### **Challenge 4: Repeat Nobel Prize Winners**

Some laureates have won the Nobel Prize **more than once**.

#### **Approach**

* Identify duplicate entries in the `full_name` column
* Use `duplicated(..., keep=False)` to capture all repeat winners

#### **Findings**

* There are **6 repeat Nobel Prize winners**
* Only **4 of them are individuals**
* Notable examples:

  * **Marie Curie** (Physics and Chemistry)
  * **Linus Carl Pauling** (Chemistry and Peace)
  * **International Committee of the Red Cross** (awarded 3 times)

---

### **Challenge 5: Number of Prizes per Category**

Questions explored:

* How many prize categories exist?
* Which categories have the most and fewest awards?

#### **Approach**

* Use `.nunique()` to count categories
* Use `.value_counts()` to count prizes per category
* Create a Plotly bar chart
* Apply the **Aggrnyl** color scale without displaying a color axis

#### **Findings**

* There are **6 Nobel Prize categories**
* Categories with the **most prizes**:

  * Medicine
  * Physics
* Category with the **fewest prizes**:

  * Economics

---

### **Challenge 6: The Economics Prize**

The relatively small number of Economics prizes raises questions.

#### **Finding**

* The **Economics Prize was first awarded in 1969**
* This explains why it has far fewer total awards compared to other categories that began in **1901**

---

### **Challenge 7: Male vs Female Winners by Category**

Analyse whether gender imbalance varies by category.

#### **Approach**

* Group data by `category` and `sex`
* Use `.groupby()` with `.agg(count)`
* Create a **split bar chart** using Plotly with `color=sex`

#### **Findings**

* Gender imbalance is largest in:

  * Physics
  * Chemistry
  * Economics
* Women are **relatively better represented** in:

  * Literature
  * Peace
  * Medicine

Split bar charts provide a **much more granular view** than overall percentages and reveal important category-level differences.
