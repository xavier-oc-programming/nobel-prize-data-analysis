### **Challenge 3: Type Conversions**

#### **Convert `birth_date` to Datetime**

```python
df_data.birth_date = pd.to_datetime(df_data.birth_date)
```

This enables proper time-based analysis later in the project.

---

#### **Create Percentage Share Column**

The `prize_share` column is stored as a fraction (e.g. `"1/2"`). Convert it into a numeric percentage.

```python
separated_values = df_data.prize_share.str.split('/', expand=True)
numerator = pd.to_numeric(separated_values[0])
denominator = pd.to_numeric(separated_values[1])

df_data['share_pct'] = numerator / denominator
```
