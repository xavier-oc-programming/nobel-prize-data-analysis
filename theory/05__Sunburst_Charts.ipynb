### **Overview**

This section explores where Nobel Prize–winning research actually takes place. Instead of focusing on countries alone, the analysis drills down into **research organisations**, **cities**, and their **geographic hierarchy**, highlighting patterns of concentration and dispersion in global scientific discovery.

### **Challenge 1: Top Research Organisations**

Many Nobel laureates are affiliated with universities, laboratories, or research organisations at the time of their discoveries. The goal is to identify which institutions are associated with the highest number of Nobel Prizes.

**Key questions:**

* Which organisations make up the top 20?
* How many Nobel Prize winners are affiliated with institutions such as the University of Chicago and Harvard University?

**Approach:**

* Count how often each `organization_name` appears.
* Select the top 20 organisations.
* Sort values to control bar chart order.

```python
top20_orgs = df_data.organization_name.value_counts()[:20]
top20_orgs.sort_values(ascending=True, inplace=True)
```

**Visualisation:**

* Horizontal Plotly bar chart
* X-axis: number of prizes
* Y-axis: institution name
* Colour by prize count

```python
org_bar = px.bar(
    x=top20_orgs.values,
    y=top20_orgs.index,
    orientation='h',
    color=top20_orgs.values,
    color_continuous_scale=px.colors.sequential.haline,
    title='Top 20 Research Institutions by Number of Prizes'
)

org_bar.update_layout(
    xaxis_title='Number of Prizes',
    yaxis_title='Institution',
    coloraxis_showscale=False
)

org_bar.show()
```

### **Challenge 2: Research Cities**

Each research organisation is located in a specific city. This challenge examines whether certain cities act as global hotspots for scientific discovery.

**Key questions:**

* Where is the number one hotspot for discoveries worldwide?

**Approach:**

* Select and sort the top 20 cities.

```python
top20_org_cities = df_data.organization_city.value_counts()[:20]
top20_org_cities.sort_values(ascending=True, inplace=True)
```

This challenge contrasts **where discoveries are made** with **where laureates were born**.

### **Challenge 4: Sunburst Chart of Research Locations**

Scientific discovery follows a geographic hierarchy: **country → city → organisation**. A sunburst chart is ideal for visualising this structure and identifying concentration patterns.

**Approach:**

* Group data by:

  * `organization_country`
  * `organization_city`
  * `organization_name`
* Count prizes per organisation.

```python
country_city_org = df_data.groupby(
    ['organization_country', 'organization_city', 'organization_name'],
    as_index=False
).agg({'prize': pd.Series.count})

country_city_org = country_city_org.sort_values('prize', ascending=False)
```

**Visualisation:**

```python
sunburst = px.sunburst(
    country_city_org,
    path=['organization_country', 'organization_city', 'organization_name'],
    values='prize',
    title='Where Do Discoveries Take Place?'
)

sunburst.update_layout(coloraxis_showscale=False)
sunburst.show()
```

**Key observations:**

* France shows extreme centralisation, with most Nobel-affiliated institutions in Paris.
* Germany’s discoveries are spread across multiple cities.
* The UK is dominated by Cambridge and London.
* The United States combines high volume with wide geographic distribution across many research hubs.

---

### **Key Takeaways**

* Research impact is far more geographically concentrated than laureate birthplaces.
* Some countries favour strong central hubs, while others distribute research across cities.
* The sunburst chart reveals structural differences in national research ecosystems.
* High prize counts do not necessarily imply geographic concentration.
