## **4)—A Choropleth Map and the Countries with the Most Prizes**

### **Overview**

This section analyses how Nobel Prizes are distributed across countries. The focus is on ranking countries by total prizes, visualising this distribution on a world map, breaking prizes down by category, and analysing how country dominance has changed over time.

---

### **Challenge 1: Top 20 Country Ranking**

**Goal:** Identify which countries have won the most Nobel Prizes.

* Create a DataFrame called `top20_countries` with:

  * `birth_country_current`
  * `prize` (total number of prizes won)

**Key consideration for country columns:**

* `birth_country`: includes countries that no longer exist (e.g. Soviet Union).
* `organization_country`: excludes individual laureates.
* **`birth_country_current`** is the least problematic and most consistent choice.

**Approach:**

* Group by `birth_country_current`
* Count prizes
* Sort by total prize count
* Select the top 20 countries

**Visualisation:**

* Create a **horizontal Plotly bar chart**
* X-axis: number of prizes
* Y-axis: country
* Colour bars by prize count
* Hide colour scale for clarity

**Observation:**

* The United States leads by a very large margin.
* The UK and Germany follow in second and third place.

---

### **Challenge 2: Choropleth Map**

**Goal:** Display global prize distribution on a world map.

**Requirements:**

* Use ISO 3-letter country codes.
* Colour countries by total prize count.
* Display country name on hover.

**Approach:**

* Group data by `birth_country_current` and `ISO`
* Count prizes
* Use `px.choropleth()` with:

  * `locations='ISO'`
  * `color='prize'`
  * A sequential colour scale (e.g. `matter`)

**Outcome:**

* Clear global contrast between high- and low-prize countries.
* Strong dominance by the United States and Western Europe.

---

### **Challenge 3: Country Bar Chart with Prize Category**

**Goal:** Break down each country’s total prizes by category.

**Steps:**

* Group data by:

  * `birth_country_current`
  * `category`
* Count prizes per category per country.
* Merge this result with `top20_countries` to retain total prize counts.
* Rename columns:

  * `cat_prize`: prizes per category
  * `total_prize`: total prizes per country
* Sort by `total_prize`.

**Visualisation:**

* Horizontal Plotly bar chart
* X-axis: `cat_prize`
* Y-axis: country
* Colour by `category`

**Insights:**

* The United States dominates **Economics**.
* Germany outperforms the UK in **Physics**.
* France has more prizes than Germany in **Peace** and **Literature**.
* The US has more prizes in **Physics or Medicine alone** than France has in total.

---

### **Challenge 4: Prizes by Country Over Time**

**Goal:** Analyse how national dominance has changed historically.

**Steps:**

* Count prizes by:

  * `birth_country_current`
  * `year`
* Sort by year.
* Compute cumulative prize counts per country.

**Visualisation:**

* Plotly **line chart**
* X-axis: year
* Y-axis: cumulative number of prizes
* Colour lines by country

**Key observations:**

* Before World War II, Nobel Prizes were largely European.
* After World War II, the United States rises sharply and becomes dominant.
* The UK and Germany swap relative positions during the 1970s–1990s.
* Sweden remains consistently high, possibly reflecting home-country bias.
* The prize becomes increasingly global in the last 40 years.

---

### **Key Takeaways**

* Country analysis depends heavily on column choice and assumptions.
* The United States dominates overall prize counts, especially post-WWII.
* Category-level breakdowns reveal strengths not visible in totals alone.
* Time-based analysis shows a clear shift from European to global dominance.
